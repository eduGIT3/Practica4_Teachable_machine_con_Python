````python
import tensorflow as tf
from keras.models import load_model
from PIL import Image, ImageOps
import numpy as np

class_names = open("/labels.txt", "r").readlines()
data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
image = Image.open("/incidente.png").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index]
confidence_score = prediction[0][index]

# Print prediction and confidence score
print("Class:", class_name[2:], end="")
print("Confidence Score:", confidence_score)
````

````python
# implementa la clasificación con la imagen que pertenece claramente a una categoría
# implementa la clasificación con la imagen que pertenece claramente a una categoría
image_path = "/botella.jpg"
image = Image.open(image_path).convert("RGB")
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index]
confidence_score = prediction[0][index]

print(f"--- Análisis de {image_path} ---")
if confidence_score < 0.95:
  print("Objeto desconocido no se puede saber con exactitud que objeto es")
else:
  print("Clase predicha:", class_name[2:].strip())
  print("Puntuación de Confianza:", confidence_score)
````

**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



Para utilizar el modelo exportado, primero se carga el archivo keras_model.h5 en el entorno de Python empleando Keras. Antes de poder realizar una predicción, es estrictamente necesario adaptar las imágenes de entrada al formato exacto que espera la red neuronal. Este proceso consiste en abrir la imagen, redimensionarla y recortarla a un tamaño fijo de 224 X 224 píxeles, y aplicar una normalización matemática sobre sus colores. Una vez preparada, la imagen se procesa con la función predict(), la cual devuelve tanto la categoría más probable como su porcentaje de confianza asociado.   

En cuanto a su comportamiento, el modelo resulta ser extremadamente preciso al evaluar imágenes claras de objetos con los que fue entrenado (como la botella), alcanzando niveles de confianza que rozan el 100%. Sin embargo, presenta una limitación importante ante objetos ambiguos o que no pertenecen a su catálogo (como el vaso o el pez): al tener que repartir la probabilidad entre sus clases conocidas, tiende a forzar una predicción errónea con aparente seguridad. Para contrarrestar esta debilidad, se ha implementado un filtro mediante código que exige un umbral mínimo de seguridad del 95%. Gracias a esta regla, cuando el modelo duda o la puntuación es inferior a ese límite, el sistema es capaz de frenar el error y clasificar correctamente la entrada como un "objeto desconocido"
